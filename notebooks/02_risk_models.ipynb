# 02 Risk Models

Run from repository root after `pip install -e .` and `python scripts/download_data.py`. These notebooks use actual provider data; no synthetic returns.

```
from adaptive_portfolio.data import read_prices, returns_from_prices
prices = read_prices('data/prices_stooq.csv')
rets = returns_from_prices(prices)

from adaptive_portfolio.risk import covariance
import pandas as pd
print(pd.DataFrame(covariance(rets),index=rets.columns,columns=rets.columns).round(4))
print(rets.corr().round(2))
```
